### For taking images of my chessboard

```python
import cv2
import os
import re

# Create folder to save images
save_dir = "captured_images"
os.makedirs(save_dir, exist_ok=True)

# --------- NEW: Determine starting image index ----------
existing_files = os.listdir(save_dir)

image_numbers = []

for file in existing_files:
    match = re.match(r"image_(\d+)\.jpg", file)
    if match:
        image_numbers.append(int(match.group(1)))

if image_numbers:
    img_count = max(image_numbers) + 1
else:
    img_count = 0
# --------------------------------------------------------

# Open default camera
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Error: Could not open camera.")
    exit()

print(f"Starting from image index: {img_count}")
print("Press 's' to save image")
print("Press 'q' to quit")

while True:
    ret, frame = cap.read()
    
    if not ret:
        print("Failed to grab frame")
        break

    cv2.imshow("Camera", frame)

    key = cv2.waitKey(1) & 0xFF

    if key == ord('s'):
        img_name = os.path.join(save_dir, f"image_{img_count}.jpg")
        cv2.imwrite(img_name, frame)
        print(f"Saved {img_name}")
        img_count += 1

    elif key == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
